# Correlation and Variable Relationships

**Question:** Is the size of a property a stronger predictor of price than its location?

In Lesson 2, the charts showed that bigger properties tend to cost more.
In this notebook I measure that relationship with numbers, and check if it is the same in every state.

## 1. Setup and loading the data

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define the path to the clean CSV file
data_path = "data/mexico-real-estate-combined-clean.csv"

# Load the data
data = pd. read_csv(data_path)

# Verify the load
print(f"Dimensions: {data.shape[0]} rows and {data.shape[1]} columns.")
data.head()
```

## 2. Correlation between area and price (whole country)

```python
# Calculate correlation between 'area_m2' and 'price_usd'
corr_price_area = (
    data['area_m2']
    .corr(data['price_usd'])
)

print(f"The correlation between Area and Price is: {corr_price_area:.4f}")
```

## 3. Correlation matrix and heatmap

```python
import numpy as np

# Select only numerical columns
# Compute the correlation matrix
corr_matrix = (
    data
    .select_dtypes("number") # to select the numeric data only
    .corr()
)

corr_matrix
```

```python
# Create a figure
fig, ax = plt.subplots()

# Create the heatmap using Seaborn
# annot=True adds the numbers to the boxes
# cmap="coolwarm" creates a nice Red (hot/positive) to Blue (cold/negative) spectrum
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", ax=ax)

ax.set_title("Correlation Matrix of Real Estate Data")
plt.show()
```

## 4. Correlation by state

The country-wide number can hide big differences between states.
I build the calculation one step at a time, then put it together.

```python
(
    data
    .head()
)
```

```python
(
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
)
```

```python
(
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
    .corr()
    .head() # <-- to only view the top 5 rows.
)
```

```python
(
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
    .corr()
    .iloc[:, 1]
    .head(10) # <-- to only view the top 10 rows
)
```

```python
(
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
    .corr()
    .iloc[:, 1]
    .xs("area_m2", level=1)
    .head() # <-- to only view the top 5 rows
)
```

```python
(
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
    .corr()
    .iloc[:, 1]
    .xs("area_m2", level=1)
    .sort_values(ascending=False)
    .head()   # <-- to only view the top 5 rows
)
```

### The full calculation (all states)

```python
# Calculate correlation by state

correlations_by_state = (
    data
    .groupby("state")
    [["area_m2", "price_usd"]]
    .corr()                          # Build correlation matrix per state
    .iloc[:, 1]                      # Keep only the price_usd column
    .xs("area_m2", level=1)          # Extract one correlation value per state
    .sort_values(ascending=False)    # Rank strongest to weakest
)

print(correlations_by_state.head(5))
print("\n... and the bottom 5 ...\n")
print(correlations_by_state.tail(5))
```

## 5. Correlation by property type

```python
import numpy as np

# Calculate correlation by property_type
correlations_by_prop = (
    data
    .groupby("property_type")   # groupby `"property_type"`
    [["area_m2", "price_usd"]]    # we select by the `"area_m2"` and `"price_usd"` columns
    .corr()
    .iloc[:, 1]
    .xs("area_m2", level=1) # extracts the `"area_m2"` correlation value from the inner index level.
)

print(correlations_by_prop)
```

## 6. New column: price per square meter

Total price depends on size. Dividing price by area lets me compare properties of different sizes.

```python
# Create new column
df = (
    data
    .assign(price_per_m2 = lambda x: x["price_usd"] / x["area_m2"])
)

df.head()
```

## 7. Are larger properties cheaper per square meter?

```python
fig, ax = plt.subplots()

# Scatter plot
ax.scatter(x=df["area_m2"], y=df["price_per_m2"],)

ax.set_title("Are larger properties cheaper per unit?")
ax.set_xlabel("Area [sq meters]")
ax.set_ylabel("Price per m2 [USD/m2]")

plt.show()
```
